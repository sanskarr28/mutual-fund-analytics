```python
import sqlite3
import pandas as pd
import os

print("1. Connecting to the Database...")
conn = sqlite3.connect('../data/db/bluestock_mf.db')

# Ensure the output directory exists
os.makedirs('../data/powerbi_export', exist_ok=True)

# List of the tables we need to extract for Power BI
tables_to_export = [
    'dim_fund', 
    'fact_nav', 
    'fact_transactions'
]

print("2. Exporting Tables to CSV...")
for table in tables_to_export:
    print(f"   ► Extracting {table}...")
    try:
        df = pd.read_sql_query(f"SELECT * FROM {table}", conn)
        export_path = f"../data/powerbi_export/{table}.csv"
        df.to_csv(export_path, index=False)
        print(f"     ✅ Saved successfully!")
    except Exception as e:
        print(f"     ❌ Error extracting {table}: {e}")

conn.close()
print("\n🎉 Extraction Complete! Your files are ready for Power BI in the 'data/powerbi_export' folder.")
```
