```python
import pandas as pd

# The '../' steps out of 'notebooks', then goes into 'data/raw'
fund_master = pd.read_csv('../data/raw/01_fund_master.csv')

print("--- UNIQUE FUND HOUSES ---")
print(fund_master['fund_house'].unique(), "\n")

print("--- UNIQUE CATEGORIES ---")
print(fund_master['category'].unique(), "\n")

print("--- UNIQUE SUB-CATEGORIES ---")
print(fund_master['sub_category'].unique(), "\n")

print("--- UNIQUE RISK GRADES ---")
print(fund_master['risk_category'].unique(), "\n")
```

```python
# 1. Load the NAV history dataset
nav_history = pd.read_csv('../data/raw/02_nav_history.csv')

# Note: If you get a KeyError here, change 'scheme_code' to match 
# the exact column names in your datasets (e.g., 'Scheme_Code' or 'Scheme Code')
master_col = 'amfi_' 
nav_col = 'amfi_code'

# 2. Extract unique codes into Python Sets for easy comparison
master_codes = set(fund_master[master_col].unique())
nav_codes = set(nav_history[nav_col].unique())

# 3. Find any codes that are in the master list but MISSING from the NAV history
missing_codes = master_codes - nav_codes

# 4. Print the Data Quality Summary
print("========================================")
print("      DATA QUALITY SUMMARY REPORT")
print("========================================")
print(f"Total unique schemes in Fund Master: {len(master_codes)}")
print(f"Total unique schemes in NAV History: {len(nav_codes)}")

if len(missing_codes) == 0:
    print("\n✅ VALIDATION PASSED: All scheme codes in Fund Master exist in NAV History.")
else:
    print(f"\n⚠️ WARNING: Found {len(missing_codes)} scheme codes in Fund Master that are MISSING from NAV History.")
    print("First 10 Missing Codes:", list(missing_codes)[:10])
```
